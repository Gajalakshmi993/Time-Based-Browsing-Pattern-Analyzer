```python
import pandas as pd
from urllib.parse import urlparse

# -----------------------------
# Load Data
# -----------------------------
history = pd.read_csv("history_last_5d_100k.csv")
domain_map = pd.read_csv("domain_category_map.csv")

# -----------------------------
# Clean column names
# -----------------------------
history.columns = history.columns.str.strip().str.lower()
domain_map.columns = domain_map.columns.str.strip().str.lower()

print("History columns:", history.columns)
print("Domain map columns:", domain_map.columns)

# -----------------------------
# Fix URL column (IMPORTANT)
# -----------------------------
# adjust if needed
if 'url' not in history.columns:
    print("⚠️ URL column not found. Available columns:", history.columns)

# -----------------------------
# Extract domain safely
# -----------------------------
def extract_domain(url):
    try:
        if pd.isna(url):
            return None
        netloc = urlparse(url).netloc
        if netloc == "":
            return None
        if netloc.startswith("www."):
            netloc = netloc[4:]
        return netloc.lower()
    except:
        return None

history['domain'] = history['domain'].astype(str).str.lower().str.strip()

# -----------------------------
# DO NOT DROP EVERYTHING ❗
# -----------------------------
print("Total rows:", len(history))
print("Valid domains:", history['domain'].notna().sum())

# Only drop if safe
history = history[
    history['domain'].notna() & (history['domain'] != '') & (history['domain'] != 'nan')]

# -----------------------------
# Clean domain_map
# -----------------------------
def clean_domain(d):
    if pd.isna(d):
        return d
    d = d.lower().strip()
    if d.startswith("www."):
        d = d[4:]
    return d

history['domain'] = history['domain'].apply(clean_domain)
domain_map['domain'] = domain_map['domain'].apply(clean_domain)
# -----------------------------
# Merge
# -----------------------------
data = pd.merge(history, domain_map, on='domain', how='left')

# -----------------------------
# Fill category safely
# -----------------------------
if 'category_y' in data.columns:
    data['category'] = data['category_y'].combine_first(data.get('category_x'))
else:
    data['category'] = data.get('category_x')

# Fill missing
data['category'] = data['category'].fillna('Others')

# Drop extra columns
data = data.drop(columns=[col for col in ['category_x', 'category_y'] if col in data.columns])

# -----------------------------
# FINAL CHECK
# -----------------------------
print("Final dataset shape:", data.shape)

if data.empty:
    print("❌ STILL EMPTY → issue in URL or column names")
else:
    print("✅ Data ready for analysis")
```

##Sessionization

```python
# Ensure timestamp exists
data['timestamp'] = pd.to_datetime(data['timestamp_local'], errors='coerce')

# Sort by time
data = data.sort_values('timestamp')

# Time difference between rows
data['time_diff'] = data['timestamp'].diff().dt.total_seconds()

# New session if gap > 15 mins (900 sec)
data['new_session'] = (data['time_diff'] > 900).astype(int)

# Assign session id
data['session_id'] = data['new_session'].cumsum()
```

```python
session_summary = data.groupby('session_id').agg({
    'timestamp': ['min', 'max', 'count'],
    'domain': 'nunique',
    'category': lambda x: x.value_counts().index[0]
})

session_summary.columns = ['start_time', 'end_time', 'events', 'unique_domains', 'top_category']

# Session duration
session_summary['duration_min'] = (
    (session_summary['end_time'] - session_summary['start_time'])
    .dt.total_seconds() / 60
)

print(session_summary.head())
```

```python
data['category_shift'] = data['category'] != data['category'].shift()

switching = data.groupby('session_id')['category_shift'].sum()

session_summary['category_switches'] = switching
```

##RAM correlation

```python
ram_log['timestamp'] = pd.to_datetime(ram_log['timestamp'], errors='coerce')

# Sort both
data = data.sort_values('timestamp')
ram_log = ram_log.sort_values('timestamp')

# Merge nearest timestamps
merged = pd.merge_asof(
    data,
    ram_log,
    on='timestamp',
    direction='nearest'
)
```

```python
ram_stats = merged.groupby('session_id').agg({
    'ram_usage': ['mean', 'max']
})

ram_stats.columns = ['avg_ram', 'peak_ram']

# Merge with session summary
session_summary = session_summary.merge(ram_stats, on='session_id', how='left')

print(session_summary.head())
```

##Pattern Discovery

```python
from sklearn.preprocessing import StandardScaler

features = session_summary[['duration_min', 'events', 'unique_domains', 'category_switches', 'avg_ram']]

features = features.fillna(0)

scaler = StandardScaler()
X = scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
session_summary['cluster'] = kmeans.fit_predict(X)
```

```python
cluster_summary = session_summary.groupby('cluster').mean()

print(cluster_summary)
```

```python
def label_cluster(row):
    if row['duration_min'] > 30 and row['avg_ram'] > 50:
        return "Work Session"
    elif row['category_switches'] > 10:
        return "Distracted"
    else:
        return "Light Usage"

session_summary['label'] = session_summary.apply(label_cluster, axis=1)
```
